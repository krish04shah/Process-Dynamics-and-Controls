```python
from importnb import Notebook
with Notebook():
    from che5 import sim, pid, TF1, TF2, shift

import numpy as np
from plotly.subplots import make_subplots
import plotly.io as pio
pio.templates.default = "plotly_dark"
import jax
import jax.numpy as jnp
jax.config.update("jax_enable_x64", True)

from plotly.subplots import make_subplots

from sympy.abc import s
from sympy import exp, Symbol, simplify
import scipy as sp
import scipy.signal as sig
```

```python
q = 100 #L/min
ca_in = 1 #mol/L
T_in = 350 #K
V = 100 #L
p = 1000 #g/L
C = 0.239 #J/gK
negHr = 5e4 #J/mol
ER = 8750 #K
k0 = 7.2e10 #1/min
UA = 5e4 #J/minK
Tc_ss = 300 #K
ca_ss = 0.5 #mol/L
T_ss = 350 #K
t_end = 10 #min
```

```python
def rhs(t,v):
  Tc = 305
  ca, T = v
  k = k0*exp(-ER/T)
  dca = (q/V)*(ca_in - ca) - k*ca
  dT = ((p*q*C)*(T_in - T) - UA*(T - Tc) + (negHr)*(V*k*ca))/(p*V*C)
  return np.array([dca, dT])
```

```python
res = sp.integrate.solve_ivp(rhs, (0, t_end), (ca_ss, T_ss), method = 'Radau', dense_output = True)
```

```python
t_plot = np.linspace(0, t_end, 200)
fig = make_subplots(rows = 1, cols = 2)
fig.add_scatter(x = t_plot, y = res.sol(t_plot)[0], row = 1, col = 1, name = 'ca')
fig.add_scatter(x = t_plot, y = res.sol(t_plot)[1], row = 1, col = 2, name = 'T')
```
