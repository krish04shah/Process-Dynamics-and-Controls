```python
import numpy as np
from plotly.subplots import make_subplots
from scipy.integrate import solve_ivp
```

```python
qin_ss = 0.5 #m3/s
A = 1.2 #m2
A_out = 0.2 #m2
g = 9.81 #m/s2

h_ss = ((qin_ss/(A_out))**2)/(2*g)

h3sp = h_ss*1.1

Kc = 6 # Controller Gain

def rhs(t, v):
  h1, h2, h3 = v
  qin0 = qin_ss + Kc*(h3sp - h3) # Proportional only controller - P-controller
  q1, q2, q3 = A_out*np.sqrt(2*g*h1), A_out*np.sqrt(2*g*h2), A_out*np.sqrt(2*g*h3)
  return [((qin0 - q1)/A),((q1 - q2)/A),((q2 - q3)/A)]
```

```python
tend = 1000.
res = solve_ivp(rhs, [0, tend], [h_ss]*3, dense_output=True, method='Radau')
res
```

```python
tplot = np.linspace(0, tend, 1000)
h1, h2, h3 = res.sol(tplot)
fig = make_subplots()
fig.add_scatter(x=tplot, y=h1, mode = 'lines', name = 'h1')
fig.add_scatter(x=tplot, y=h2, mode = 'lines', name = 'h2')
fig.add_scatter(x=tplot, y=h3, mode = 'lines', name = 'h3')
```
